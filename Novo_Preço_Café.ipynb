### 1- Buscando base de dados com API yfinance

```python
"""

Instalacao das dependencias:
    pip install yfinance pandas

Como rodar:
    - Em script normal (terminal):   python preco_do_cafe.py
    - Em notebook (Jupyter/VS Code): from preco_do_cafe import main
                                      df = main()
"""

# -- 1. IMPORTS ----------------------------------------------------------
from pathlib import Path         # caminhos de arquivo multiplataforma
import pandas as pd              # manipulacao e exportacao dos dados
import yfinance as yf            # acesso a series historicas do Yahoo Finance
import matplotlib.pyplot as plt  # biblioreca para criação de gráficos
```

```python
# -- 2. CONFIGURACAO ------------------------------------------------------
OUTPUT_DIR = Path("dados_agro")
OUTPUT_DIR.mkdir(exist_ok=True)

# Quantos anos de historico buscar.
ANOS_HISTORICO = 4

# Ticker do contrato futuro de cafe arabica na ICE (Yahoo Finance).
# Cotado em US cents/lb (centavos de dolar por libra-peso).
TICKER_CAFE = "KC=F"


# -- 3. CALCULAR A DATA DE INICIO -----------------------------------------
def data_inicio(anos: int) -> str:
    """
    Calcula a data de `anos` atras a partir de hoje, no formato
    'AAAA-MM-DD' esperado pelo parametro `start` do yfinance.
    """
    hoje = pd.Timestamp.today()
    inicio = hoje - pd.DateOffset(years=anos)
    return inicio.strftime("%Y-%m-%d")

```

```python
# -- 4. FUNCAO DE DOWNLOAD -------------------------------------------------
def baixar_preco_cafe(anos: int = ANOS_HISTORICO) -> pd.DataFrame:
    """
    Busca a serie diaria de preco do cafe (futuro ICE, ticker KC=F) para
    os ultimos `anos` anos e retorna um DataFrame do pandas com os dados.

    Diferente da versao anterior (baseada em agrobr/cepea), o yfinance
    nao precisa de await/asyncio: a chamada e sincrona.
    """
    inicio = data_inicio(anos)
    print(f"Baixando preco diario do cafe ({TICKER_CAFE}) desde {inicio}...")

    ticker = yf.Ticker(TICKER_CAFE)
    try:
        df = ticker.history(start=inicio, interval="1d", auto_adjust=False)
    except Exception as e:
        raise RuntimeError(
            "Falha ao consultar o Yahoo Finance (rede indisponivel, proxy "
            "bloqueando o dominio, ou instabilidade temporaria do servico). "
            f"Erro original: {e}"
        ) from e

    if df.empty:
        raise RuntimeError(
            "Nao foi possivel obter dados do Yahoo Finance para "
            f"'{TICKER_CAFE}'. Verifique a conexao com a internet, se o "
            "ticker ainda e valido e se o pacote yfinance esta atualizado "
            "('pip install -U yfinance')."
        )

    # Organiza e traduz as colunas para um formato mais claro em pt-BR.
    df = df.reset_index()
    df = df.rename(columns={
        "Date": "data",
        "Open": "abertura",
        "High": "maxima",
        "Low": "minima",
        "Close": "fechamento",
        "Volume": "volume",
    })
    df["data"] = pd.to_datetime(df["data"]).dt.tz_localize(None).dt.date
    df["produto"] = "cafe_arabica_futuro"
    df["unidade"] = "USD cents/lb"
    df["fonte"] = "Yahoo Finance (ticker KC=F, futuro ICE)"

    colunas = ["data", "abertura", "maxima", "minima", "fechamento", "volume",
               "produto", "unidade", "fonte"]
    colunas = [c for c in colunas if c in df.columns]
    df = df[colunas].sort_values("data").reset_index(drop=True)

    caminho = OUTPUT_DIR / "preco_cafe_ice_4anos.csv"
    df.to_csv(caminho, index=False, encoding="utf-8-sig")
    print(f"  OK -> {caminho} ({len(df)} linhas, {len(df.columns)} colunas)")

    return df

```

```python
# -- 5. FUNCAO PRINCIPAL ----------------------------------------------------
def main() -> pd.DataFrame:
    df = baixar_preco_cafe()
    print("\nPrevia dos dados:")
    print(df.head())
    print("...")
    print(df.tail())
    return df

```

```python
# -- 6. GRAFICO -------------------------------------------------------------
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Paleta e specs de uma serie temporal unica (linha fina, sem legenda --
# um unico titulo ja identifica o que esta plotado).
COR_LINHA = "#2a78d6"        # azul (cor padrao para serie unica)
COR_GRADE = "#e1e0d9"        # cinza claro, hairline (so no eixo Y)
COR_EIXO = "#c3c2b7"         # cinza para spines/baseline
COR_TEXTO_PRIM = "#0b0b0b"   # tinta primaria (titulo)
COR_TEXTO_SEC = "#52514e"    # tinta secundaria (labels dos eixos)
COR_TEXTO_MUTED = "#898781"  # tinta apagada (ticks)
COR_SUPERFICIE = "#fcfcfb"   # fundo do grafico


def plotar_preco_cafe(df: pd.DataFrame, salvar: bool = True) -> None:
    """
    Gera um grafico de linha do preco diario de fechamento do cafe
    (coluna 'fechamento', em US cents/lb) ao longo do periodo baixado.
    """
    df_plot = df.copy()
    df_plot["data"] = pd.to_datetime(df_plot["data"])

    fig, ax = plt.subplots(figsize=(11, 4.5), facecolor=COR_SUPERFICIE)
    ax.set_facecolor(COR_SUPERFICIE)

    ax.plot(
        df_plot["data"], df_plot["fechamento"],
        color=COR_LINHA, linewidth=1.6,
        solid_capstyle="round", solid_joinstyle="round",
    )

    ax.set_title(
        f"Preco diario do cafe ({TICKER_CAFE}, futuro ICE) "
        f"- ultimos {ANOS_HISTORICO} anos",
        color=COR_TEXTO_PRIM, fontsize=13, fontweight="bold", pad=12,
    )
    ax.set_xlabel("Data", color=COR_TEXTO_SEC)
    ax.set_ylabel("US cents / lb", color=COR_TEXTO_SEC)

    # So gradeado horizontal (hairline) -- reduz ruido visual numa serie
    # diaria longa; o eixo X ja fica marcado pelos ticks de data.
    ax.grid(True, axis="y", color=COR_GRADE, linewidth=0.8)
    ax.set_axisbelow(True)

    for spine in ("top", "right"):
        ax.spines[spine].set_visible(False)
    for spine in ("left", "bottom"):
        ax.spines[spine].set_color(COR_EIXO)

    ax.tick_params(colors=COR_TEXTO_MUTED, labelsize=9)
    locator = mdates.AutoDateLocator()
    ax.xaxis.set_major_locator(locator)
    ax.xaxis.set_major_formatter(mdates.ConciseDateFormatter(locator))

    fig.tight_layout()

    if salvar:
        caminho_grafico = OUTPUT_DIR / "grafico_preco_cafe_ice_4anos.png"
        fig.savefig(caminho_grafico, dpi=150, facecolor=fig.get_facecolor())
        print(f"Grafico salvo em: {caminho_grafico}")

    plt.show()
```

### 4- Análise de variáveis - análise descritiva, decomposição de componentes

4.1 - Preparando a série para decomposição (frequência regular)

```python
"""
A série original só tem dias úteis, mas com furos irregulares
(feriados que não caem sempre no mesmo dia). Para decompor em
tendência/sazonalidade/resíduo o statsmodels precisa de um índice
de frequência REGULAR, para ser possivel definir um periodo.
"""

import numpy as np

df_cafe['data'] = pd.to_datetime(df_cafe['data'])
serie = df_cafe.set_index('data')['fechamento'].asfreq('B')  # força frequência de dia útil

print(f'Buracos criados pelo asfreq (feriados): {serie.isna().sum()} de {len(serie)} dias úteis')

"""
Preenchendo os feriados assumindo que o preço entre dois pregões variam de
forma suave quando o furo é de 1-2 dias.
"""

serie = serie.interpolate(method='time')
print(f'Nulos após interpolação: {serie.isna().sum()}')
```

4.2 - Análise descritiva

```python
# Estatísticas descritivas da série já tratada
print(serie.describe())

log_ret = np.log(serie / serie.shift(1)).dropna()
print('\nRetornos log diários:')
print(log_ret.describe())
print(f"Assimetria (skew): {log_ret.skew():.3f}")
print(f"Curtose (excess): {log_ret.kurtosis():.3f}")  # > 0 = caudas mais pesadas que a normal
```

```python
fig, axes = plt.subplots(2, 1, figsize=(11, 6))

axes[0].plot(serie.index, serie.values, lw=0.8, color='#2a78d6')
axes[0].set_title('Preço de fechamento - série tratada (frequência diária útil)')
axes[0].set_ylabel('USD cents/lb')

axes[1].plot(log_ret.index, log_ret.values, lw=0.5, color='darkred')
axes[1].axhline(0, color='black', lw=0.5)
axes[1].set_title('Retornos log diários (proxy de volatilidade)')

plt.tight_layout()
plt.show()
```

### Interpretação

Ao observarmos os números e os gráficos, alguns pontos são mais discrepantes sobre o comportamento do café:

*   **O preço está sempre me movimento:** Como esperado para uma commodity, a série de preços não é estável (não estacionária). Ela não gira em torno de uma média fixa e a sua variação muda conforme o tempo passa, apresentando tendências claras de alta ou baixa.
*   **Volatilidade:** Quando olhamos para os retornos diários, percebemos que eles oscilam perto de zero, mas o tamanho dessas oscilações não é constante. Tivemos períodos mais tranquilos (2023 até metade de 2024) e momentos de grande agitação (2025 em diante).
*   **Eventos Extremos:** A curtose acima de zero nos diz que o café tem caudas pesadas. Isso significa que movimentos bruscos de preço (aqueles que saem muito do comum) acontecem com mais frequência do que se o mercado seguisse uma distribuição normal perfeita. É um sinal de alerta para a etapa de testes de normalidade que faremos adiante.

4.3 - Decomposição de componentes

```python
from statsmodels.tsa.seasonal import STL

# STL exige um período sazonal definido. Para dado diário não faz muito
# sentido impor uma sazonalidade "de calendário" (não há razão para o
# preço do café seguir um padrão fixo por dia da semana). Por isso,
# agregamos para frequência semanal antes de decompor, testando
# período = 52 (uma sazonalidade anual, compatível com o ciclo de
# safra/florada do café).

serie_semanal = serie.resample('W').mean()

stl = STL(serie_semanal, period=52, robust=True)
resultado = stl.fit()

fig = resultado.plot()
fig.set_size_inches(11, 8)
plt.tight_layout()
plt.show()
```

### Análise decomposição

Ao separar o preço em tendência, sazonalidade e resíduo, conseguimos notar alguns detalhes importantes:

*   **Tendência Dominante:** O gráfico de tendência confirma o que vimos visualmente: o café teve uma subida forte entre 2024 e 2025, com um recuo mais recente. É o motor principal do preço.
*   **Sazonalidade:** O modelo até aponta um padrão que se repete a cada 52 semanas, mas com apenas 4 anos de dados, é difícil dizer se isso é uma sazonalidade real da safra ou apenas o modelo tentando achar um padrão em meio ao ruído. Para o trabalho, é mais seguro dizer que existe um indício de sazonalidade, mas que a série ainda é curta para uma afirmação categórica.
*   **Resíduo:** Onde o gráfico de resíduos oscila mais forte é justamente onde tivemos os picos de volatilidade. Isso mostra que ainda existe informação ali (autocorrelação) que a simples decomposição não pegou, e é exatamente isso que vamos investigar nos próximos passos de validação estatística.
